## Загрузка данных

Вам дан датасет о ценах на жильев с характеристиками этого жилья: `sales_dataset.csv`.

```python
# @title
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
file_path = '/sales_dataset.csv'
sales_data = pd.read_csv(file_path)

sales_data.head(), sales_data.info()
```

## График 1

Построить точечный график зависимости переменной `Area_in_Marla` от `price` с учётом третей переменной `purpose`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(data=sales_data, x="price", y="Area_in_Marla", hue="purpose", alpha=0.6)
plt.title("Scatter Plot: Area_in_Marla vs. Price by Purpose", fontsize=14)
plt.xlabel("Price (in millions)", fontsize=12)
plt.ylabel("Area in Marla", fontsize=12)
plt.legend(title="Purpose", loc="upper right")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

Вывод: стоимость аренды жилья увеличивается в зависимости от площади, однако варьируется в пределах небольшого диапазона и остается значительно ниже цен на жилье для продажи. Цены на продажу жилья слабо коррелируют с площадью: в основном предлагаются объекты площадью до 1000 марла по цене до 250 000 000 рупий.

## График 2

Постройте 4 графика на одной фигуре, как расположить координатные плоскости - это на ваше усмотрение.

Это должны быть гистограммы для следующих переменных:
- `property_type`
- `city`
- `purpose`
- `bedrooms`

Каждый график должен быть своего цвета, также задайте цвет отличный от белого для бекграунда всей фигуры.

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.patch.set_facecolor('lightgrey')

sns.histplot(sales_data["property_type"], ax=axes[0, 0], color="skyblue", kde=False)
axes[0, 0].set_title("Distribution of Property Types", fontsize=12)

sns.histplot(sales_data["city"], ax=axes[0, 1], color="lightgreen", kde=False)
axes[0, 1].set_title("Distribution of Cities", fontsize=12)
axes[0, 1].tick_params(axis='x', rotation=45)

sns.histplot(sales_data["purpose"], ax=axes[1, 0], color="coral", kde=False)
axes[1, 0].set_title("Distribution of Purposes", fontsize=12)

sns.histplot(sales_data["bedrooms"], ax=axes[1, 1], color="gold", kde=False)
axes[1, 1].set_title("Distribution of Bedrooms", fontsize=12)

plt.tight_layout()
plt.show()
```

Вывод:  
1) На рынке недвижимости чаще всего представлены частные дома, реже — квартиры, и наименее распространены пентхаусы.  
2) Наибольшее количество предложений жилья сосредоточено в Карачи, тогда как в Фейсалабаде их значительно меньше.  
3) Объем предложений жилья для продажи почти в три раза превышает количество объектов, доступных для аренды.  
4) Наиболее популярны дома с тремя спальнями, в то время как варианты с 10 и более спальнями встречаются редко (менее 100 предложений).

## График 3

Визуализируйте матрицу корреляции в виде heatmap.

Все должно быть ЧИТАБЕЛЬНО!

```python
numeric_data = sales_data.select_dtypes(include=['number'])
correlation_matrix = numeric_data.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True, square=True)
plt.title("Correlation Matrix Heatmap", fontsize=16)
plt.tight_layout()
plt.show()
```

Вывод:  
1) Координаты местоположения жилья (широта и долгота) в большинстве случаев демонстрируют взаимосвязь, что может указывать на концентрацию объектов в определенных районах.  
2) Наблюдается зависимость между месяцем и сезоном, что, вероятно, отражает сезонные колебания активности на рынке недвижимости.  
3) Также выявлена связь между количеством спален и ванных комнат, что свидетельствует о стандартных планировках жилья в зависимости от его размеров.

## График 5

Постройте heatmap для отображения пропусков в данных.

```python

plt.figure(figsize=(10, 6))
sns.heatmap(sales_data.isnull(), cbar=False, cmap="viridis")
plt.title("Heatmap of Missing Data", fontsize=14)
plt.tight_layout()
plt.show()
```

Пропусков данных нет ни в одной категории

## График 6

Постройте график среднего значения переменной `price` в зависимости от переменной `Season`.

```python
avg_price_by_season = sales_data.groupby("Season")["price"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=avg_price_by_season, x="Season", y="price", palette="muted")
plt.title("Average Price by Season", fontsize=14)
plt.xlabel("Season", fontsize=12)
plt.ylabel("Average Price", fontsize=12)
plt.tight_layout()
plt.show()
```

Выводы:  

На основе анализа графиков видно, что средняя цена на недвижимость варьируется в зависимости от сезона. Наибольшее значение цены наблюдается во второй сезон, тогда как в четвертый сезон цены достигают минимальных значений. Таким образом, наиболее выгодным временем для продажи или сдачи недвижимости является второй сезон, а для покупки или заключения договоров аренды — четвертый сезон.
